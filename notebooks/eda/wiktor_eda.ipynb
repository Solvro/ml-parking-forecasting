## **Ogarnianie datasetu:**



```python
# Uzupełnię NaN'y dla Polinki, jest to parking całodobowy więc przedział będzie 00:00-23:59

df_parkings['open_hour'] = df_parkings['open_hour'].fillna('00:00:00')
df_parkings['close_hour'] = df_parkings['close_hour'].fillna('23:59:59')

# Zamiana dla wszystkich wartości w tych kolumnach aby były na pewno godzinowe fajne

df_parkings['open_hour'] = pd.to_datetime(df_parkings['open_hour'], format='%H:%M:%S').dt.time
df_parkings['close_hour'] = pd.to_datetime(df_parkings['close_hour'], format='%H:%M:%S').dt.time
```

```python
# Skoro formaty danych i NaN'y ogarnięte to teraz przydałoby się znormalizować miare wolnych miejsc, wprowadzę nową kolumne occupancy_rate w %.

# Narazie złącze dwa najprostsze do połączenia datasety: df_prakings i df_availabilities,  w jednego dataframe'a z sensownymi kolumnami (df_full)

df_full = df_availabilities.merge(df_parkings[['id', 'name', 'places']], left_on='parking_id', right_on='id', suffixes=('', '_parking'))
df_full['occupancy_rate'] = (df_full['places'] - df_full['spaces_left']) / df_full['places'] * 100
df_full.drop(columns=['created_at', 'updated_at'], inplace = True)


df_full.head()
```

```python
# Sprawdzam jeszcze czy teraz w złączonej tabelce nie ma anomalii i czy nie pojawia się ujemne occupancy rate:

# Wtedy kiedy procentowe zajecie miejsc jest ujemne oznacza to ze parking jest pusty, odwrotnie - parking pełny.
anomalies_low = df_full[df_full['occupancy_rate'] < 0]
anomalies_high = df_full[df_full['occupancy_rate'] > 100]

print(f"=== ANOMALIE ===")
print(f"Wartości ujemne: {len(anomalies_low)} wierszy")
print(f"Wartości powyżej 100%: {len(anomalies_high)} wierszy")

# Sprawdzam który parking jest najbardziej zepsuty
print("\nTopka wadliwych parkingów:")
all_anomalies = pd.concat([anomalies_low, anomalies_high])
print(all_anomalies['name'].value_counts())

print("\nPrzykładowe błędne wiersze:")
display(all_anomalies[['measured_at', 'name', 'places', 'spaces_left', 'occupancy_rate']].head())
```

```python
# Clippuje, wszystko poniżej 0 zamień na 0, wszystko powyżej 100 na 100
df_full['occupancy_rate'] = df_full['occupancy_rate'].clip(lower=0, upper=100)

# Skoro zmieniliśmy % zajętości, musimy zaktualizować liczbę wolnych miejsc,
df_full['spaces_left'] = df_full['places'] - (df_full['occupancy_rate'] / 100 * df_full['places'])

# Zaokrąglamy do pełnych aut
df_full['spaces_left'] = df_full['spaces_left'].round().astype(int)

# Sprawdzonko jak to teraz wygląda
print(f"Min occupancy: {df_full['occupancy_rate'].min()}%")
print(f"Max occupancy: {df_full['occupancy_rate'].max()}%")
```

```python
# Przed doklejaniem kolejnych feature'ów sortuje dataset
df_full = df_full.sort_values('measured_at')
display(df_full.head())
```

## **Łączenie datasetów:**




### **Użytkownicy w sks'ie:**

```python
# Sortujemy obie tabele po czasie (wymóg aby użyć metody merge_asof)
df_full = df_full.sort_values('measured_at')
df_sksusers = df_sksusers.sort_values('external_timestamp')

# Wybieramm tylko (moim zdaniem) przydatne cechy z sks_userów
sks_subset = df_sksusers[['external_timestamp', 'active_users', 'moving_average_21']]
```

```python
# Używam merge_asof, bo dla każdego pomiaru parkingu znajduje liczbę użytkowników z najbliższego momentu w czasie

df_full = pd.merge_asof(
    df_full,                    # Tabela LEWA (główna)
    sks_subset,                 # Tabela PRAWA (dostawiana)
    left_on='measured_at',      # Czas w parkingu
    right_on='external_timestamp', # Czas w SKS
    direction='nearest',        # Szukaj najbliższej wartości
    tolerance=pd.Timedelta('1 hour') # Maksymalnie 1h różnicy
)

df_full.drop(columns=['external_timestamp'], inplace=True)

display(df_full.head())
```

```python
# Zobaczmy, czy mamy jakieś NaNy (gdzie nie udało się dopasować SKS)
missing_sks = df_full['active_users'].isna().sum()
print(f"Brakujących danych SKS: {missing_sks}")
```

```python
# Trzeba uzupełnić i active_users jak i moving_average_21
# Używam interpolacji, method='time' uwzględnia odległości czasowe

df_full['active_users'] = df_full.set_index('measured_at')['active_users'].interpolate(method='time').values
df_full['moving_average_21'] = df_full.set_index('measured_at')['moving_average_21'].interpolate(method='time').values

# Jeśli braki są na samym początku lub końcu pliku, interpolacja tam nie sięgnie. Używam ffill (wypełnij ostatnią znaną) i bfill (wypełnij następną znaną)
df_full['active_users'] = df_full['active_users'].ffill().bfill()
df_full['moving_average_21'] = df_full['moving_average_21'].ffill().bfill()

# Szybki check
print(f"Pozostałe braki SKS: {df_full['active_users'].isna().sum()}, braki w wygładzeniu: {df_full['moving_average_21'].isna().sum()}")
```

###**Eventy:**

```python
# Najpierw sprawdzam jak sie mają eventowe daty do dat parkingowych

print("Zakres Parkingi:", df_full['measured_at'].min(), "do", df_full['measured_at'].max())
print("Zakres Eventy:  ", df_events['start'].min(), "do", df_events['end'].max())  # wtf 😭
```

```python
# Trzeba teraz 'wyciąć' eventy matchujące datą do parkingowych danych

parking_start = df_full['measured_at'].min()
parking_end = df_full['measured_at'].max()

matching_events = df_events[(df_events['end'] >= parking_start) & (df_events['start'] <= parking_end)].copy()

print(f"Liczba eventów matchujących: {len(matching_events)}")

# Przygotowanie do włożenia eventów w df_full

df_full['is_event'] = 0

for idx, row in matching_events.iterrows():
    mask = (df_full['measured_at'] >= row['start']) & (df_full['measured_at'] <= row['end'])
    df_full.loc[mask, 'is_event'] = 1
```

```python
# Sprawdzam pokrycie eventowe 2025 roku
event_coverage = df_full['is_event'].mean() * 100
print(f"\nProcent czasu z aktywnym eventem: {event_coverage:.2f}%")
```

### **Dodawanie dodatkowych cech i robienie porządku w finalnym datasecie:**

```python
# Podsumowując, po dodaniu całości otrzymaliśmy:

df_full.head(10)
```

```python
# Będę pracował teraz na df_model

df_model = df_full.copy()
df_model = df_model.sort_values(by=['name', 'measured_at'])

# Usunięcie zbędnych kolumn
cols_to_drop = ['id', 'parking_id', 'id_parking', 'spaces_left']
df_model = df_model.drop(columns=[c for c in df_model.columns if c in cols_to_drop])

# Wyciągnięcie czasowych cech
df_model['hour'] = df_model['measured_at'].dt.hour
df_model['day_of_week'] = df_model['measured_at'].dt.dayofweek
df_model['month'] = df_model['measured_at'].dt.month
df_model['is_weekend'] = (df_model['day_of_week'] >= 5).astype(int)

# Dodanie lagów

df_model['lag_15m'] = df_model.groupby('name')['occupancy_rate'].shift(15)
df_model['lag_1h'] = df_model.groupby('name')['occupancy_rate'].shift(60)
df_model['lag_24h'] = df_model.groupby('name')['occupancy_rate'].shift(1440)
df_model = df_model.dropna()

# One-hot na nazwy parkingów
df_model = pd.get_dummies(df_model, columns=['name'], prefix='parking')

# Szybki check jak to się prezentuje
print("=== GOTOWE DANE DO MODELU ===")
print(df_model.info())
display(df_model.head())
```

## **Piękne wizualizacje:**



```python
# Wybieramy tylko numeryczne kolumny, które mają sens do korelacji
cols_to_corr = ['occupancy_rate', 'active_users', 'is_event', 'hour', 'day_of_week', 'month', 'lag_15m', 'lag_1h', 'lag_24h']

# Obliczamy korelację
corr_matrix = df_model[cols_to_corr].corr()

# 3. Rysowanie
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True,fmt=".2f", cmap='coolwarm',vmin=-1, vmax=1,linewidths=1.5)

plt.title('Macierz Korelacji:')
plt.show()
```

```python
day_map = {0: 'Poniedziałek', 1: 'Wtorek', 2: 'Środa', 3: 'Czwartek', 4: 'Piątek', 5: 'Sobota', 6: 'Niedziela'}
df_model['day_name'] = df_model['day_of_week'].map(day_map)
days_order = list(day_map.values())

weekly_heatmap = df_model.groupby(['day_name', 'hour'])['occupancy_rate'].mean().unstack()
weekly_heatmap = weekly_heatmap.reindex(days_order)

# Rysowanie
plt.figure(figsize=(12, 6))
sns.heatmap(weekly_heatmap, cmap='RdYlGn_r', annot=False)

plt.title('Tygodniowy cykl życia parkingu (Średnia zajętość %)')
plt.xlabel('Godzina')
plt.ylabel('Dzień tygodnia')
plt.show()
```

```python
parking_cols = [c for c in df_model.columns if c.startswith('parking_')]
df_model['temp_name'] = df_model[parking_cols].idxmax(axis=1).str.replace('parking_', '')

monthly_heatmap = df_model.groupby(['temp_name', 'month'])['occupancy_rate'].mean().unstack()

# Rysowanie
plt.figure(figsize=(12, 6))
sns.heatmap(monthly_heatmap, cmap='magma_r', annot=True, fmt=".2f")

plt.title('Sezonowość: Średnia zajętość (%) w poszczególnych miesiącach')
plt.xlabel('Miesiąc')
plt.ylabel('Parking')
plt.show()
```
